# [Virtual Environment]

> **一行要約: 仮想環境とは、特定のプログラム専用の独立した実行環境を意味し、プログラムごとに適切なパッケージバージョンなどを適用することで、安全な作業環境を構築してくれます。**

以下のコードは、仮想環境を作成する例です。

```python
%%bash

# プロジェクトフォルダを作成後、該当フォルダへ移動
mkdir basic-project && cd basic-project

# 仮想環境（venv）の作成 | py -3.11 -m venv .venv（Pythonの特定バージョンを指定。仮想環境名の先頭に「.」を付けて隠しフォルダとして管理）
python -m venv venv

# 仮想環境の有効化（アクティベート）（※ Mac/Linuxでは activate の代わりに bin を使用）
source venv/Scripts/activate

pip install pandas
pip list

deactivate
```

仮想環境フォルダにはインストールしたパッケージがすべて含まれているため、.gitignore に指定してGitHubにはプッシュしません。

代わりに requirements.txt ファイルへパッケージ情報を保存してアップロードします。

以下は requirements.txt の例です。

```python
%%bash

# 既存のプロジェクトフォルダへ移動後、パッケージ一覧を抽出
cd ../ml-analysis
source .venv/Scripts/activate
pip freeze > requirements.txt
deactivate

# 新規プロジェクトフォルダへ移動後、仮想環境を作成
cd ..
mkdir team-project && cd team-project
python -m venv venv
source venv/Scripts/activate

# 保存したファイルをもとに、パッケージを一括同期（インストール）
pip install -r ../ml-analysis/requirements.txt
```

以下は、仮想環境がおかしくなった場合に、Linuxの削除コマンドでリセットする方法です。

```python
%%bash

有効化されている場合は、まず無効化
deactivate

# venv フォルダを強制削除（Linuxコマンド）
rm -rf venv
# (PowerShell - Remove-Item -Recurse -Force venv)

# クリーンな状態で新規作成および有効化
python -m venv venv
source venv/Scripts/activate
```

以下は、すでに仮想環境を有効化（実行）している状態でPythonのバージョンを変更する方法です。

```python
%%bash

# 該当のPythonバージョンがなく、インストールが必要な場合は先に以下のコードを実行します：
# winget install Python.Python.3.11 (PowerShell/ GitBash(Windows)) - 仮想環境を有効化したままインストールしても問題ありません

# 既存の venv を削除
Remove-Item -Recurse -Force venv

# Python 3.11 を指定して venv を作成
py -3.11 -m venv venv

# 仮想環境を有効化
.\venv\Scripts\Activate.ps1

# バージョンを確認
python --version
```
